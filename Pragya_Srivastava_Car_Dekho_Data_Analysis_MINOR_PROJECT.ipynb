```python
# Q11: First-owner vehicles

first_owner_count = (df['Owner'] == 0).sum()

print("Number of First-Owner Vehicles:", first_owner_count)

owner_counts = df['Owner'].value_counts().sort_index()

plt.figure(figsize=(8, 5))
plt.bar(owner_counts.index.astype(str), owner_counts.values)
plt.xlabel("Owner Category")
plt.ylabel("Number of Vehicles")
plt.title("Vehicle Distribution by Number of Previous Owners")
plt.show()
```

```python
# Q12: Calculate depreciation

df['Depreciation'] = (
    df['Present_Price'] -
    df['Selling_Price']
)

df['Depreciation_Percentage'] = (
    df['Depreciation'] /
    df['Present_Price']
) * 100

# Most depreciated
most_depreciated = df.loc[
    df['Depreciation'].idxmax()
]

# Least depreciated
least_depreciated = df.loc[
    df['Depreciation'].idxmin()
]

print("MOST DEPRECIATED VEHICLE")
print(most_depreciated)

print("\nLEAST DEPRECIATED VEHICLE")
print(least_depreciated)

# Top 10 depreciation
top_depreciation = df.sort_values(
    'Depreciation',
    ascending=False
).head(10)

plt.figure(figsize=(12, 6))
plt.bar(
    top_depreciation['Car_Name'],
    top_depreciation['Depreciation']
)
plt.xlabel("Vehicle")
plt.ylabel("Depreciation (Lakhs)")
plt.title("Top 10 Most Depreciated Vehicles")
plt.xticks(rotation=75)
plt.show()
```

```python
# Q 13: Extract and standardize Brand names from Car_Name
brand_map = {
    'Ritz': 'Maruti',
    'Sx4': 'Maruti',
    'Ciaz': 'Maruti',
    'Wagon': 'Maruti',
    'Swift': 'Maruti',
    'Vitara': 'Maruti',
    'S': 'Maruti',
    'Alto': 'Maruti',
    'Ertiga': 'Maruti',
    'Dzire': 'Maruti',
    'Ignis': 'Maruti',
    '800': 'Maruti',
    'Baleno': 'Maruti',
    'Omni': 'Maruti',
    'Fortuner': 'Toyota',
    'Innova': 'Toyota',
    'Corolla': 'Toyota',
    'Etios': 'Toyota',
    'Camry': 'Toyota',
    'Land': 'Toyota',
    'Activa': 'Honda',
    'I20': 'Hyundai',
    'Grand': 'Hyundai',
    'I10': 'Hyundai',
    'Eon': 'Hyundai',
    'Xcent': 'Hyundai',
    'Elantra': 'Hyundai',
    'Creta': 'Hyundai',
    'Verna': 'Hyundai',
    'City': 'Honda',
    'Brio': 'Honda',
    'Amaze': 'Honda',
    'Jazz': 'Honda',
    'Royal': 'Royal Enfield',
}

df['Brand'] = (
    df['Car_Name']
    .apply(lambda x: str(x).split()[0].title())
    .replace(brand_map)
)

# Calculate Depreciation Percentage: ((Present Price - Selling Price) / Present Price) * 100
df['Depreciation_Percentage'] = (
    (df['Present_Price'] - df['Selling_Price']) / df['Present_Price']
) * 100

# Group by Brand and compute average depreciation
brand_dep = (
    df.groupby('Brand')['Depreciation_Percentage']
    .agg(['mean', 'count'])
    .reset_index()
)
brand_dep = brand_dep[brand_dep['count'] >= 3].sort_values(
    by='mean'
)  # Filter for statistical significance

# --- Plot Diagram ---
plt.figure(figsize=(10, 6))
sns.barplot(data=brand_dep, x='mean', y='Brand', palette='viridis')
plt.title(
    'Average Cost Depreciation Percentage by Brand',
    fontsize=14,
    fontweight='bold',
)
plt.xlabel('Average Depreciation (%)', fontsize=12)
plt.ylabel('Vehicle Brand', fontsize=12)

for index, value in enumerate(brand_dep['mean']):
  plt.text(value + 1, index, f'{value:.1f}%', va='center')

plt.xlim(0, 100)
plt.tight_layout()
plt.show()

# Print Answer
print('Least Depreciated Brands (Percentage Wise):')
print(brand_dep[['Brand', 'mean']].rename(columns={'mean': 'Avg_Depreciation_%'}))
```

```python
# Q14: Correlation analysis

correlation = df[
    [
        'Year',
        'Selling_Price',
        'Present_Price',
        'Kms_Driven',
        'Owner',
        'Depreciation'
    ]
].corr()

print("Correlation Matrix:")
display(correlation.round(2))

plt.figure(figsize=(9, 7))

sns.heatmap(
    correlation,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title("Correlation Heatmap of Factors Affecting Depreciation")
plt.show()
```

```python
# Q15: Relationship between age, kilometers and selling price

# Calculate vehicle age
df['Vehicle_Age'] = 2018 - df['Year']

# Correlations
age_corr = df['Vehicle_Age'].corr(df['Selling_Price']
)

kms_corr = df['Kms_Driven'].corr(
    df['Selling_Price']
)

print("Correlation between Vehicle Age and Selling Price:",
      round(age_corr, 3))

print("Correlation between KMs Driven and Selling Price:",
      round(kms_corr, 3))


# Age vs Selling Price
plt.figure(figsize=(9, 6))
plt.scatter(
    df['Vehicle_Age'],
    df['Selling_Price'],
    alpha=0.6
)
plt.xlabel("Vehicle Age (Years)")
plt.ylabel("Selling Price (Lakhs)")
plt.title("Vehicle Age vs Selling Price")
plt.show()


# KMs vs Selling Price
plt.figure(figsize=(9, 6))
plt.scatter(
    df['Kms_Driven'],
    df['Selling_Price'],
    alpha=0.6
)
plt.xlabel("Kilometers Driven")
plt.ylabel("Selling Price (Lakhs)")
plt.title("Kilometers Driven vs Selling Price")
plt.show()
```

```python
# Q16: Vehicles manufactured after 2014

new_vehicles = df[
    df['Year'] > 2014
]

print("Number of Vehicles Manufactured After 2014:",
      len(new_vehicles))

print("\nYear-wise Distribution:")
print(
    new_vehicles['Year']
    .value_counts()
    .sort_index()
)

year_new = new_vehicles['Year'].value_counts().sort_index()

plt.figure(figsize=(8, 5))
plt.bar(
    year_new.index.astype(str),
    year_new.values
)
plt.xlabel("Manufacturing Year")
plt.ylabel("Number of Vehicles")
plt.title("Vehicles Manufactured After 2014")
plt.show()
```

```python
# Q17: Extract two-wheelers

two_wheeler_keywords = [
    'activa',
    'apache',
    'avenger',
    'bajaj',
    'bullet',
    'classic',
    'discover',
    'duke',
    'fz',
    'honda cb',
    'karizma',
    'ktm',
    'maestro',
    'mojo',
    'passion',
    'pulsar',
    'r15',
    'rd',
    'royal enfield',
    'splendor',
    'street',
    'thunderbird',
    'tvs',
    'um renegade',
    'wagon r'
]

def is_two_wheeler(name):
    name = name.lower()
    return any(keyword in name for keyword in two_wheeler_keywords)

df['Vehicle_Type'] = df['Car_Name'].apply(
    lambda x: 'Two-Wheeler'
    if is_two_wheeler(x)
    else 'Car'
)

two_wheelers = df[
    df['Vehicle_Type'] == 'Two-Wheeler'
].copy()

cars = df[
    df['Vehicle_Type'] == 'Car'
].copy()

print("Two-Wheeler Records:", len(two_wheelers))
print("Car Records:", len(cars))

# Pie chart
type_counts = df['Vehicle_Type'].value_counts()

plt.figure(figsize=(7, 7))
plt.pie(
    type_counts.values,
    labels=type_counts.index,
    autopct='%1.1f%%',
    startangle=90
)
plt.title("Cars vs Two-Wheelers")
plt.show()
```

```python
# Q18: Oldest bike

oldest_bike = two_wheelers.loc[
    two_wheelers['Year'].idxmin()
]

print("Oldest Bike:")
display(oldest_bike.to_frame().T)

# Bike year distribution
bike_years = two_wheelers['Year'].value_counts().sort_index()

plt.figure(figsize=(10, 5))
plt.bar(
    bike_years.index.astype(str),
    bike_years.values
)
plt.xlabel("Manufacturing Year")
plt.ylabel("Number of Bikes")
plt.title("Two-Wheeler Distribution by Manufacturing Year")
plt.xticks(rotation=45)
plt.show()
```

```python
# Q19: Newest bike

newest_bike = two_wheelers.loc[
    two_wheelers['Year'].idxmax()
]

print("Newest Bike:")
display(newest_bike.to_frame().T)

print("\nNewest Bike Year:",
      newest_bike['Year'])
```

```python
# Q20: Most sold bike

bike_counts = two_wheelers[
    'Car_Name'
].value_counts()

print("Most Sold Bike:", bike_counts.index[0])
print("Number of Records:", bike_counts.iloc[0])

# Top 10 bikes
top_bikes = bike_counts.head(10)

plt.figure(figsize=(10, 6))
plt.barh(
    top_bikes.index[::-1],
    top_bikes.values[::-1]
)

plt.xlabel("Number of Records")
plt.ylabel("Bike")
plt.title("Top 10 Most Sold Bikes")
plt.show()
```

```python
# Q21: Exceptional two-wheeler deals

two_wheelers['Price_Retention'] = (
    two_wheelers['Selling_Price'] /
    two_wheelers['Present_Price']
) * 100

# Highest price retention
best_bike_deal = two_wheelers.loc[
    two_wheelers['Price_Retention'].idxmax()
]

print("Best Two-Wheeler Deal:")
display(best_bike_deal.to_frame().T)

# Scatter plot
plt.figure(figsize=(10, 6))

plt.scatter(
    two_wheelers['Present_Price'],
    two_wheelers['Selling_Price'],
    alpha=0.7
)

plt.xlabel("Present Price (Lakhs)")
plt.ylabel("Selling Price (Lakhs)")
plt.title("Present Price vs Selling Price - Two-Wheelers")

plt.show()
```

```python
# Q22: Extract only cars

cars = df[
    df['Vehicle_Type'] == 'Car'
].copy()

print("Number of Car Records:", len(cars))

display(cars.head())

# Pie chart
type_counts = df['Vehicle_Type'].value_counts()

plt.figure(figsize=(7, 7))
plt.pie(
    type_counts.values,
    labels=type_counts.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title("Cars vs Two-Wheelers")
plt.show()
```

```python
# Q23: Oldest car

oldest_car = cars.loc[
    cars['Year'].idxmin()
]

print("Oldest Car:")
display(oldest_car.to_frame().T)

# Car manufacturing year distribution
car_years = cars['Year'].value_counts().sort_index()

plt.figure(figsize=(10, 5))
plt.bar(
    car_years.index.astype(str),
    car_years.values
)

plt.xlabel("Manufacturing Year")
plt.ylabel("Number of Cars")
plt.title("Cars by Manufacturing Year")
plt.xticks(rotation=45)

plt.show()
```

```python
# Q24: Newest car

newest_car = cars.loc[
    cars['Year'].idxmax()
]

print("Newest Car:")
display(newest_car.to_frame().T)

print("\nNewest Car Manufacturing Year:",
      newest_car['Year'])
```

```python
# Q25: Exceptional car deal

cars['Price_Retention'] = (
    cars['Selling_Price'] /
    cars['Present_Price']
) * 100

# Best car deal
best_car_deal = cars.loc[
    cars['Price_Retention'].idxmax()
]

print("Best Car Deal:")
display(best_car_deal.to_frame().T)

# Scatter plot
plt.figure(figsize=(10, 6))

plt.scatter(
    cars['Present_Price'],
    cars['Selling_Price'],
    alpha=0.6
)

plt.xlabel("Present Price (Lakhs)")
plt.ylabel("Selling Price (Lakhs)")
plt.title("Present Price vs Selling Price - Cars")

plt.show()
```
